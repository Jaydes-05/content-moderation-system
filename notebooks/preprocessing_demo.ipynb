# Preprocessing Module Demo

This notebook demonstrates how to use the `src/preprocessing` module for text cleaning.

```python
import sys
sys.path.insert(0, '../src')

from preprocessing import clean_text, preprocess_pipeline
import pandas as pd
```

## 1. Basic Usage

```python
# Sample toxic comment
toxic = "You're an IDIOT!!! Check this: https://spam.com 😡😡"

print("Original:")
print(toxic)
print("\nCleaned:")
print(clean_text(toxic))
```

## 2. Pipeline Modes

```python
sample = "Check OUT this LINK: https://spam.com !!! You're WRONG 😡"

print(f"Original:\n  {sample}\n")
print(f"Default mode:\n  {preprocess_pipeline(sample, mode='default')}\n")
print(f"ML mode (for TF-IDF):\n  {preprocess_pipeline(sample, mode='ml')}\n")
print(f"Transformer mode (for BERT):\n  {preprocess_pipeline(sample, mode='transformer')}")
```

## 3. Apply to DataFrame

```python
# Load dataset
df = pd.read_csv('../data/processed/train_cleaned.csv')

print(f"Dataset shape: {df.shape}")
df.head(3)
```

```python
# Apply preprocessing (default mode)
df['comment_cleaned'] = df['comment_text'].apply(lambda x: preprocess_pipeline(x, mode='default'))

# Show before/after
print("Before:")
print(df['comment_text'].iloc[0][:200])
print("\nAfter:")
print(df['comment_cleaned'].iloc[0][:200])
```

```python
# Check for empty comments after cleaning
empty_count = (df['comment_cleaned'].str.len() == 0).sum()
print(f"Empty comments after cleaning: {empty_count} ({empty_count/len(df)*100:.2f}%)")
```

## 4. Compare Modes on Real Data

```python
# Take a sample comment
sample_idx = 100
original = df['comment_text'].iloc[sample_idx]

print("Original:")
print(original)
print("\n" + "="*70 + "\n")

for mode in ['default', 'ml', 'transformer']:
    cleaned = preprocess_pipeline(original, mode=mode)
    print(f"Mode '{mode}':")
    print(cleaned)
    print()
```

## 5. Save Preprocessed Data

```python
# Save with default preprocessing
output_path = '../data/processed/train_preprocessed.csv'
df.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
```
